# 실습과제: 환자 응급도 분류하기

응급실에 내원한 환자의 생체징후를 확인하여 응급도를 분류하고 적절한 처치 방향을 안내하는 프로그램을 작성해보세요.

## 요구사항

1. **응급도 분류 기준**
   - 즉시 처치: 의식이 없거나 수축기혈압 ≥ 180 또는 체온 ≥ 39.0
   - 응급 처치: 통증이 8점 이상 또는 호흡곤란
   - 준응급: 통증이 4-7점 또는 발열(37.5-38.9도)
   - 비응급: 위 조건에 해당하지 않는 경우

2. **처치 방향 및 대기시간**
   - 즉시 처치: "즉시 응급처치" / 0분
   - 응급 처치: "응급 처치 후 정밀검사" / 15분
   - 준응급: "약물 처치 후 전문의 진료" / 30분
   - 비응급: "일반 진료" / 60분

3. **출력 형식**
```
=== 환자 응급도 분류 결과 ===
환자명: 김환자 (45세)
생체징후: 혈압 160/90, 체온 38.2도, 통증 3점
의식상태: 의식 있음, 호흡곤란: 호흡 없음

🟡 분류 결과: 준응급
📋 처치 방향: 약물 처치 후 전문의 진료
⏰ 예상 대기시간: 30분
```

### 힌트

- `if-elif-else`문으로 응급도를 분류하세요
- `and`/`or` 연산자로 복합 조건을 처리하세요
- `not` 연산자로 의식 상태를 확인하세요
- f-string을 사용해 결과를 예쁘게 출력하세요

## 💻 구현해보세요

````python
# 기본 데이터 - 환자 정보
patient = {
    "name": "김환자",                      # 환자 이름
    "age": 45,                            # 환자 나이 (세)
    "systolic_bp": 160,                   # 수축기 혈압 (mmHg)
    "temperature": 38.2,                  # 체온 (섭씨)
    "pain_level": 3,                      # 통증 점수 (0-10점)
    "is_conscious": True,                 # 의식 상태 (True: 의식있음, False: 의식없음)
    "has_breathing_difficulty": False     # 호흡곤란 여부 (True: 있음, False: 없음)
}
````

````python
# 여기에 코드를 작성하세요

# 1. 환자 기본 정보 출력

print("=== 환자 응급도 분류 결과 ===")

print(f"환자명: {patient['name']} ({patient['age']}세)")

print(
    f"생체징후: 혈압 {patient['systolic_bp']}/90, "
    f"체온 {patient['temperature']}도, "
    f"통증 {patient['pain_level']}점"
)

# 의식 상태 출력

consciousness_status = "의식 있음" if patient["is_conscious"] else "의식 없음"

# 호흡 상태 출력

breathing_status = "호흡 있음" if patient["has_breathing_difficulty"] else "호흡 없음"

print(f"의식상태: {consciousness_status}, 호흡곤란: {breathing_status}")

print()

# 2. 응급도 분류

# 즉시 처치 조건:
# - 의식이 없거나
# - 수축기 혈압 >= 180
# - 체온 >= 39.0

if (
    not patient["is_conscious"]
    or patient["systolic_bp"] >= 180
    or patient["temperature"] >= 39.0
):

    urgency_level = "즉시 처치"
    urgency_icon = "🔴"
    treatment = "즉시 응급처치"
    wait_time = "0분"

# 응급 처치 조건:
# - 통증 점수 >= 8
# - 호흡곤란 있음

elif patient["pain_level"] >= 8 or patient["has_breathing_difficulty"]:

    urgency_level = "응급 처치"
    urgency_icon = "🟠"
    treatment = "응급 처치 후 정밀검사"
    wait_time = "15분"

# 준응급 조건:
# - 통증 점수 4~7
# - 체온 37.5 ~ 38.9

elif (
    (patient["pain_level"] >= 4 and patient["pain_level"] <= 7)
    or
    (patient["temperature"] >= 37.5 and patient["temperature"] < 39.0)
):

    urgency_level = "준응급"
    urgency_icon = "🟡"
    treatment = "약물 처치 후 전문의 진료"
    wait_time = "30분"

# 나머지는 비응급

else:

    urgency_level = "비응급"
    urgency_icon = "🟢"
    treatment = "일반 진료"
    wait_time = "60분"

# 3. 결과 출력

print(f"{urgency_icon} 분류 결과: {urgency_level}")

print(f"📋 처치 방향: {treatment}")

print(f"⏰ 예상 대기시간: {wait_time}")
````

````python
# 위에서 작성된 코드를 다시 작성해보세요 !
# 💡 이번에는 같은 로직을 '함수' 2개로 나눠서 다시 작성합니다.
#    - classify_patient(patient): 응급도만 판단해서 결과를 돌려줌(return)  → 판단 담당
#    - print_report(patient)    : 정해진 형식으로 화면에 출력              → 출력 담당
#    왜? 판단과 출력을 나눠 두면 다른 환자에게도 그대로 재사용할 수 있고,
#        판단 결과만 따로 꺼내 assert로 테스트하기도 쉬워집니다.


# [준비] 아래 '2. 응급도 분류' 단계에서 사용할 분류 함수 (if-elif-else 활용)
def classify_patient(patient):  # 환자 정보(딕셔너리)를 받아 응급도를 판단하는 함수 정의
    """환자 정보를 받아 (아이콘, 응급도, 처치 방향, 대기시간(분)) 4개 값을 돌려줍니다."""
    bp = patient["systolic_bp"]                                 # 수축기 혈압 꺼내기 (짧은 이름에 담아 조건식을 읽기 쉽게)
    temp = patient["temperature"]                               # 체온 꺼내기
    pain = patient["pain_level"]                                # 통증 점수 꺼내기
    conscious = patient["is_conscious"]                         # 의식 여부 꺼내기 (True: 있음 / False: 없음)
    breathing_difficulty = patient["has_breathing_difficulty"]  # 호흡곤란 여부 꺼내기 (True: 있음 / False: 없음)

    # ⚠️ if-elif-else는 위에서부터 차례로 검사하다가 '처음으로 참(True)인 조건' 하나만 실행합니다.
    #    그래서 가장 위급한 '즉시 처치'를 맨 위에 둡니다. → 조건의 순서 = 우선순위
    if (not conscious) or bp >= 180 or temp >= 39.0:  # 의식이 없거나(not) 또는(or) 혈압 180 이상 또는 체온 39.0 이상이면
        urgency_icon = "🔴"                  # 응급도 아이콘
        urgency_level = "즉시 처치"           # 응급도 분류 결과
        treatment = "즉시 응급처치"            # 처치 방향
        wait_time = 0                        # 예상 대기시간(분) — 나중에 계산에도 쓸 수 있게 숫자로 저장
    elif pain >= 8 or breathing_difficulty:  # (위 조건이 거짓일 때만 검사) 통증 8점 이상 또는 호흡곤란이 있으면
        urgency_icon = "🟠"                  # 응급도 아이콘
        urgency_level = "응급 처치"           # 응급도 분류 결과
        treatment = "응급 처치 후 정밀검사"     # 처치 방향
        wait_time = 15                       # 예상 대기시간(분)
    elif (pain >= 4 and pain <= 7) or (temp >= 37.5 and temp < 39.0):  # 통증 4~7점(and) 또는(or) 발열 37.5 이상 39.0 미만(and)이면
        # 💡 문제에는 '37.5-38.9도'라고 되어 있지만 temp <= 38.9 로 쓰면 38.95도 같은 값이 빠지는 빈틈이 생겨서
        #    '39.0 미만(< 39.0)'으로 씁니다. 39.0 이상은 이미 맨 위 '즉시 처치'에서 걸러집니다.
        urgency_icon = "🟡"                  # 응급도 아이콘
        urgency_level = "준응급"              # 응급도 분류 결과
        treatment = "약물 처치 후 전문의 진료"  # 처치 방향
        wait_time = 30                       # 예상 대기시간(분)
    else:  # 위의 어떤 조건에도 해당하지 않으면
        urgency_icon = "🟢"                  # 응급도 아이콘
        urgency_level = "비응급"              # 응급도 분류 결과
        treatment = "일반 진료"               # 처치 방향
        wait_time = 60                       # 예상 대기시간(분)

    return urgency_icon, urgency_level, treatment, wait_time  # 4개 값을 튜플로 묶어서 돌려줌


def print_report(patient):  # 환자 1명의 분류 결과를 문제에서 요구한 형식 그대로 출력하는 함수
    """환자 응급도 분류 결과를 정해진 출력 형식으로 출력합니다."""

    # 1. 환자 기본 정보 출력
    print("=== 환자 응급도 분류 결과 ===")                     # 제목 줄 출력
    print(f"환자명: {patient['name']} ({patient['age']}세)")  # f-string으로 이름·나이를 끼워 넣어 출력
    diastolic_bp = patient.get("diastolic_bp", 90)          # 이완기 혈압: 딕셔너리에 없으면 문제 예시처럼 90을 기본값으로 사용
    print(  # 생체징후 한 줄 출력 (길어서 여러 줄로 나눠 씀)
        f"생체징후: 혈압 {patient['systolic_bp']}/{diastolic_bp}, "  # 혈압 (수축기/이완기)
        f"체온 {patient['temperature']}도, "                      # 체온
        f"통증 {patient['pain_level']}점"                          # 통증 점수
    )  # 괄호 안에 문자열을 나란히 쓰면 자동으로 이어 붙여져 한 줄로 출력됩니다
    consciousness_status = "의식 있음" if patient["is_conscious"] else "의식 없음"          # True면 '의식 있음', False면 '의식 없음'
    breathing_status = "호흡 있음" if patient["has_breathing_difficulty"] else "호흡 없음"  # 템플릿 출력 형식 그대로 표기 (뜻: 호흡곤란 있음/없음)
    print(f"의식상태: {consciousness_status}, 호흡곤란: {breathing_status}")  # 의식·호흡곤란 상태 출력
    print()                                                  # 빈 줄 출력 (출력 형식 맞추기)

    # 2. 응급도 분류 (if-elif-else 활용)
    urgency_icon, urgency_level, treatment, wait_time = classify_patient(patient)  # 위 함수가 돌려준 4개 값을 차례로 변수에 나눠 담기(언패킹)

    # 3. 처치 방향 및 대기시간 안내
    print(f"{urgency_icon} 분류 결과: {urgency_level}")  # 아이콘 + 응급도 출력
    print(f"📋 처치 방향: {treatment}")                  # 처치 방향 출력
    print(f"⏰ 예상 대기시간: {wait_time}분")             # 숫자 대기시간 뒤에 단위 '분'을 붙여 출력


# 실행: 맨 위에서 만든 기본 환자(김환자) 데이터로 결과 출력
print_report(patient)  # 기대 결과: 🟡 준응급 / 약물 처치 후 전문의 진료 / 30분
````

````python
# ✅ 추가 테스트: 4가지 응급도에 하나씩 해당하는 예시 환자로 함수가 맞게 동작하는지 확인
# 왜? 기본 환자 1명(준응급)만으로는 즉시 처치·응급 처치·비응급 분기가 맞는지 알 수 없기 때문입니다.
#     특히 기준선 위의 값(체온 39.0도, 37.5도 같은 '경계값')을 일부러 넣어 >=, < 를 맞게 썼는지 확인합니다.

test_cases = [  # (환자 정보, 기대하는 응급도) 쌍을 리스트에 담기
    (  # 테스트 1: 즉시 처치 예상
        {"name": "박고열", "age": 72, "systolic_bp": 150, "temperature": 39.0, "pain_level": 5,  # 환자 정보 딕셔너리
         "is_conscious": True, "has_breathing_difficulty": False},  # 체온이 정확히 39.0(경계값) + 통증 5점(준응급 조건도 해당)
        "즉시 처치",  # 39.0은 'temp >= 39.0'에 걸림 → 맨 위 조건이 먼저 검사되므로 준응급보다 우선
    ),
    (  # 테스트 2: 응급 처치 예상
        {"name": "최호흡", "age": 30, "systolic_bp": 130, "temperature": 36.8, "pain_level": 2,  # 환자 정보 딕셔너리
         "is_conscious": True, "has_breathing_difficulty": True},  # 호흡곤란만 있음
        "응급 처치",  # 'pain >= 8 or breathing_difficulty'에서 호흡곤란(True)에 걸림
    ),
    (  # 테스트 3: 준응급 예상
        {"name": "정미열", "age": 25, "systolic_bp": 120, "temperature": 37.5, "pain_level": 1,  # 환자 정보 딕셔너리
         "is_conscious": True, "has_breathing_difficulty": False},  # 체온이 정확히 37.5(발열이 시작되는 경계값)
        "준응급",  # 'temp >= 37.5'이므로 발열에 해당
    ),
    (  # 테스트 4: 비응급 예상
        {"name": "한건강", "age": 35, "systolic_bp": 118, "temperature": 37.4, "pain_level": 3,  # 환자 정보 딕셔너리
         "is_conscious": True, "has_breathing_difficulty": False},  # 체온 37.4(발열 바로 아래), 통증 3점(4점 바로 아래)
        "비응급",  # 어떤 조건에도 안 걸려서 else로 감
    ),
]

for test_patient, expected in test_cases:  # 테스트 케이스를 하나씩 꺼내 반복 (환자 정보, 기대 응급도)
    print_report(test_patient)  # 보고서를 출력해 눈으로도 확인
    _, level, _, _ = classify_patient(test_patient)  # 돌려받은 4개 값 중 응급도(두 번째)만 쓰고 나머지는 _ 로 버림
    assert level == expected, f"{test_patient['name']}: 기대값 {expected}, 실제 {level}"  # 다르면 바로 에러를 내서 알려줌
    print(f"✅ 테스트 통과: {test_patient['name']} → {level}\n")  # 통과 메시지 (\n: 다음 환자와 한 줄 띄우기)

print("🎉 4가지 응급도 테스트를 모두 통과했습니다!")  # assert가 하나라도 실패하면 이 줄은 실행되지 않습니다
````

## 핵심 정리

- **if-elif-else는 위에서부터 검사해 '처음으로 참인 조건' 하나만 실행합니다.** 그래서 조건 순서가 곧 우선순위입니다. 가장 위급한 '즉시 처치'를 맨 위에 둬야 체온 39.0도에 통증 5점인 환자(박고열)가 준응급이 아니라 즉시 처치로 분류됩니다.
- **and / or / not**: `and`는 둘 다 참이어야 참(통증 4점 이상 **그리고** 7점 이하), `or`는 하나만 참이어도 참(의식 없음 **또는** 혈압 180 이상 **또는** 체온 39.0 이상), `not`은 참·거짓을 뒤집습니다(`not conscious` → 의식이 없을 때 참).
- **경계값 처리**: 즉시 처치는 `temp >= 39.0`, 준응급 발열은 `temp >= 37.5 and temp < 39.0`으로 써서 39.0도가 정확히 한 곳(즉시 처치)에만 속하게 했습니다. `temp <= 38.9`로 쓰면 38.95도 같은 값이 준응급에서 빠져 비응급으로 잘못 분류됩니다.
